```python
import torch
import torch.nn as nn

x = torch.tensor([-3., -1., 0., 1., 3.])

activations = {
    "ReLU": nn.ReLU(),
    "LeakyReLU": nn.LeakyReLU(0.1),
    "Sigmoid": nn.Sigmoid(),
    "Tanh": nn.Tanh(),
    "SiLU": nn.SiLU(),
    "GELU": nn.GELU()
}

for name, activation in activations.items():
    print(name, activation(x).tolist())
```
